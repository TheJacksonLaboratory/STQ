```python
import os
import sys
import numpy as np
import tifffile
from tqdm import tqdm
from IPython.display import display, HTML

def setNotebookWidth(widthPercent=100):
    """Set the notebook container width in a Jupyter environment."""
    display(HTML(f"""<style>:root {{ --jp-notebook-max-width: {widthPercent}%; }}
body .container, div.container {{ width: {widthPercent}% !important; }}</style>"""))
    return

setNotebookWidth(100)

sys.path.append('../lib/')
from wsiMask import getInTissuePixelMask

sys.path.append('../utils/')
from selector import annotate_contours
```

```python
slides = ['slide_KOMP_K17745_5.ndpi',
         'slide_KOMP_K17745_10.ndpi',
         'slide_KOMP_C8389_3.ndpi',
         'slide_KOMP_C8382_1.ndpi',
         'slide_KOMP_C8389_2.ndpi',
         'slide_KOMP_K17752_10.ndpi',
         'slide_KOMP_C8382_3.ndpi']
spath = '/projects/chuang-lab/KOMP_Histopathology_Testing/Images/SlideFiles/'
imgs = [tifffile.imread(f'{spath}{s}', level=2) for s  in slides]

for i, img in tqdm(enumerate(imgs), total=len(imgs)):
    whp = (img>250).all(axis=2)
    img[whp] = np.quantile(img[~whp], 0.99)
    m = getInTissuePixelMask(img, None, kernel_size=7)
    wh = m.T==0
    imgs[i][wh, ...] = 255
```

```python
annotate_contours(samples=slides, images=imgs, canvas_max_dim=1300,
                savepath="/projects/chuang-lab/USERS/domans/dev-komp/data2/")
```
